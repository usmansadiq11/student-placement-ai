# Exploratory Data Analysis (EDA)

This notebook is used for performing exploratory data analysis on the student placement dataset. The goal is to visualize data distributions, correlations, and insights that can inform feature selection and model training.

```python
# Importing necessary libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Setting the aesthetic style of the plots
sns.set(style='whitegrid')
```

```python
# Load the dataset
data = pd.read_csv('../dataset/placement_data.csv')

# Display the first few rows of the dataset
data.head()
```

```python
# Summary statistics of the dataset
data.describe()
```

```python
# Checking for missing values
missing_values = data.isnull().sum()
missing_values[missing_values > 0]
```

```python
# Visualizing the distribution of CGPA
plt.figure(figsize=(10, 6))
sns.histplot(data['CGPA'], bins=20, kde=True)
plt.title('Distribution of CGPA')
plt.xlabel('CGPA')
plt.ylabel('Frequency')
plt.show()
```

```python
# Visualizing the correlation matrix
plt.figure(figsize=(12, 8))
correlation_matrix = data.corr()
sns.heatmap(correlation_matrix, annot=True, fmt='.2f', cmap='coolwarm', square=True)
plt.title('Correlation Matrix')
plt.show()
```

```python
# Visualizing the relationship between CGPA and placement status
plt.figure(figsize=(10, 6))
sns.boxplot(x='Placed', y='CGPA', data=data)
plt.title('CGPA vs Placement Status')
plt.xlabel('Placed')
plt.ylabel('CGPA')
plt.show()
```

## Conclusion

This exploratory data analysis provides insights into the dataset, including the distribution of CGPA, correlation between features, and the relationship between CGPA and placement status. These insights will guide the feature selection and model training processes.
